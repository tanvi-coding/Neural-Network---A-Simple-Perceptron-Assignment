#**2. Explain the basic architecture and functioning of a Perceptron. What are its limitations?**

Ans - **Perceptron Architecture and Functioning**

A Perceptron is one of the simplest types of artificial neural networks and is used for binary classification tasks. It was introduced by Frank Rosenblatt in 1957. It is designed to mimic the basic decision-making process of a biological neuron.

#Basic Architecture of a Perceptron

A perceptron consists of:

**1. Input Layer**

. Receives input features x1,x2,...,xn.

. Each input is assigned a weight that represents its importance.

**2. Weights**

. Each input has an associated weight w1,w2,...,wn.

. Weights are adjusted during training to improve predictions.

**3. Bias**

. A bias value is added to shift the decision boundary and improve flexibility.

**4. Summation Function**

. Calculates the weighted sum of inputs: z=w1x1+w2x2+...+wnxn+b

**5. Activation Function**

.The weighted sum is passed through an activation function (usually a step function).

. It produces the final output, typically 0 or 1.

#Limitations of a Perceptron

. Cannot solve non-linear problems: A single-layer perceptron can only classify data that is linearly separable. It cannot solve problems like the XOR problem.

. Limited complexity: It can handle only simple classification tasks because it has only one layer.

. Requires feature engineering: It cannot automatically learn complex features from raw data.

. Binary output limitation: Basic perceptrons are mainly designed for binary classification.

. Sensitive to data quality: Poorly scaled or noisy data can affect performance.

Both loss functions and cost functions measure how well a neural network's predictions match the actual values. They help the network learn by providing an error value that is minimized during training.

**Purpose:**

#**6. Write a Python program to implement a single-layer perceptron from scratch using NumPy to solve the logical AND gate.**
# Implementing a Single-Layer Perceptron for AND Gate using NumPy



```python
# Implementing a Single-Layer Perceptron for AND Gate using NumPy

import numpy as np

# AND gate input data
X = np.array([[0, 0],
          [0, 1],
          [1, 0],
          [1, 1]])

# Expected output
y = np.array([0, 0, 0, 1])

# Initialize weights and bias
weights = np.zeros(2)
bias = 0

# Learning rate
learning_rate = 0.1

# Number of training epochs
epochs = 10

# Step activation function
def activation_function(value):
    return 1 if value >= 0 else 0

# Training the perceptron
for epoch in range(epochs):
    for i in range(len(X)):
        # Calculate weighted sum
        z = np.dot(X[i], weights) + bias

        # Predict output
        y_pred = activation_function(z)

        # Calculate error
        error = y[i] - y_pred

        # Update weights and bias
        weights += learning_rate * error * X[i]
        bias += learning_rate * error

    print(f"Epoch {epoch+1}: Weights = {weights}, Bias = {bias}")

# Testing the perceptron
print("\nAND Gate Predictions:")
for inputs in X:
    result = activation_function(np.dot(inputs, weights) + bias)
    print(f"Input: {inputs} Output: {result}")
```

#Observation:

The single-layer perceptron successfully learns the AND gate because the AND function is linearly separable. After training, it correctly predicts the output for all four possible input combinations.

#**8. Use Keras to build and train a simple multilayer neural network on the MNIST digits dataset. Print the training accuracy.**




```python
# Build and train a simple multilayer neural network on MNIST dataset using Keras

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.datasets import mnist

# Load MNIST dataset
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Normalize pixel values (0-255) to range (0-1)
x_train = x_train / 255.0
x_test = x_test / 255.0

# Build the neural network model
model = Sequential([
Flatten(input_shape=(28, 28)),       # Convert 2D images into 1D vector
Dense(128, activation='relu'),       # Hidden layer
Dense(64, activation='relu'),        # Hidden layer
Dense(10, activation='softmax')      # Output layer for 10 classes
])

# Compile the model
model.compile(
optimizer='adam',
loss='sparse_categorical_crossentropy',
metrics=['accuracy']
)

# Train the model
history = model.fit(
x_train,
y_train,
epochs=10,
batch_size=32,
validation_data=(x_test, y_test)
)

# Evaluate the model
test_loss, test_accuracy = model.evaluate(x_test, y_test)

# Print training accuracy
print("Training Accuracy:", history.history['accuracy'][-1])

# Print test accuracy
print("Test Accuracy:", test_accuracy)
```

**Observation:**

The multilayer neural network achieves high accuracy on the MNIST handwritten digit classification task. The training accuracy is around 99%, while the test accuracy is around 97%, showing that the model learns the digit patterns effectively and generalizes well to unseen data.

#**9. Visualize the loss and accuracy curves for a neural network model trainedon the Fashion MNIST dataset. Interpret the training behavior.**


```python
import tensorflow as tf
import matplotlib.pyplot as plt
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Flatten, Dense

# Load Fashion MNIST dataset
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.fashion_mnist.load_data()

# Normalize pixel values
x_train = x_train / 255.0
x_test = x_test / 255.0

# Build neural network model
model = Sequential([
Flatten(input_shape=(28, 28)),
Dense(128, activation='relu'),
Dense(64, activation='relu'),
Dense(10, activation='softmax')
])

# Compile model
model.compile(
optimizer='adam',
loss='sparse_categorical_crossentropy',
metrics=['accuracy']
)

# Train the model
history = model.fit(
x_train,
y_train,
epochs=10,
batch_size=32,
validation_data=(x_test, y_test)
)

# Plot training and validation accuracy
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.title('Accuracy Curve')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)

# Plot training and validation loss
plt.subplot(1,2,2)
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Loss Curve')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)

plt.show()
```

#**10. You are working on a project for a bank that wants to automatically detect fraudulent transactions. The dataset is large, imbalanced, and contains structured features like transaction amount, merchant ID, and customer location. The goal is to classify each transaction as fraudulent or legitimate.**
Explain your real-time data science workflow:

● How would you design a deep learning model (perceptron or multilayer NN)?

● Which activation function and loss function would you use, and why?

● How would you train and evaluate the model, considering class imbalance?

● Which optimizer would be suitable, and how would you prevent overfitting?

```python
# Example Deep Learning Workflow for Fraud Detection

import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.optimizers import Adam
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, roc_auc_score

# Example structured transaction data
# Features: transaction_amount, merchant_id, customer_location, etc.
X = np.array([
[100, 1, 10],
[5000, 2, 20],
[50, 1, 10],
[7000, 3, 30],
[200, 2, 20],
[9000, 3, 30]
])

# Labels: 0 = Legitimate, 1 = Fraud
y = np.array([0, 1, 0, 1, 0, 1])

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
X, y, test_size=0.3, random_state=42
)

# Scale numerical features
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# Build multilayer neural network
model = Sequential([
Dense(64, activation='relu', input_shape=(X_train.shape[1],)),
Dropout(0.3),
Dense(32, activation='relu'),
Dropout(0.3),
Dense(1, activation='sigmoid')
])

# Compile model
model.compile(
optimizer=Adam(learning_rate=0.001),
loss='binary_crossentropy',
metrics=['accuracy']
)

# Handle class imbalance using class weights
class_weights = {
0: 1,
1: 3
}

# Train the model
history = model.fit(
X_train,
y_train,
epochs=20,
batch_size=16,
class_weight=class_weights,
verbose=0
)

# Evaluate the model
y_prob = model.predict(X_test)
y_pred = (y_prob > 0.5).astype(int)

print(classification_report(y_test, y_pred))
print("ROC-AUC Score:", roc_auc_score(y_test, y_prob))
```

#. Deep Learning Model Design

For a large structured banking dataset, I would use a Multilayer Neural Network (MLP) rather than a single perceptron.

Input layer:
. Takes transaction features such as amount, merchant ID, customer location, transaction time, and previous transaction behavior.

Hidden layers:
. Multiple dense layers learn complex relationships between transaction patterns.

Output layer:
. A single neuron with a probability score indicating whether a transaction is fraudulent.

A single-layer perceptron would not be suitable because fraud patterns are complex and usually not linearly separable.

#2. Activation Function and Loss Function

Hidden Layers: ReLU
. ReLU is used because it trains faster and reduces the vanishing gradient problem.

. It works well for deep neural networks.

Output Layer: Sigmoid
. Fraud detection is a binary classification problem:

. 0 → Legitimate transaction

. 1 → Fraudulent transaction

. Sigmoid produces a probability between 0 and 1.

Loss Function: Binary Cross-Entropy
. Suitable for binary classification.

. Measures the difference between predicted fraud probability and actual labels.

#3. Training and Evaluation with Class Imbalance
Fraud datasets are usually highly imbalanced because fraudulent transactions are much fewer than legitimate ones.

Approaches:

. Use class weights to give more importance to fraudulent samples.

. Apply sampling techniques such as:

. Oversampling fraud cases (SMOTE)

. Undersampling legitimate cases

. Use evaluation metrics beyond accuracy:

. Precision

. Recall

. F1-score

. ROC-AUC

. Precision-Recall AUC

Recall is especially important because missing a fraudulent transaction can cause financial loss.

#4. Optimizer Selection

Adam optimizer is suitable because:

. It combines momentum and adaptive learning rates.

. It converges faster than basic gradient descent.

. It performs well on large and complex datasets.

#5. Preventing Overfitting

To improve model generalization:

. Use Dropout layers to randomly deactivate neurons during training.

. Apply L2 regularization to penalize large weights.

. Use early stopping to stop training when validation performance stops improving.

. Monitor validation loss and use cross-validation.
